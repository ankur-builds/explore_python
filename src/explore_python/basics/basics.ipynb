```python
numbers = [10, 20, 30]

print(id(numbers))

numbers.append(40)
print(numbers)
print(id(numbers))

new_list = numbers

print(new_list)
print(id(new_list))

new_list = numbers.copy()
print(new_list)
print(id(new_list))
```

```python
import copy

backup = copy.deepcopy(numbers)
print(numbers)
print(id(numbers))

print(backup)
print(id(backup))
```

```python
def add_agent(agent, agents=[]):
    """
    Python evaluates default arguments once, when defining the function.
    That list is reused across calls.
    """
    agents.append(agent)
    return agents

print(add_agent("Researcher"))
print(add_agent("Analyst"))
print(add_agent("Reviewer"))
```

```python
# Correct Implementation
def add_agent(agent, agents=None):

    if agents is None:
        agents = []

    agents.append(agent)

    return agents

print(add_agent("Researcher"))
print(add_agent("Analyst"))
print(add_agent("Reviewer"))
```

```python
import ast
import dis

source_code = """
def calculate_total(price: float, tax_rate: float) -> float:
    return price * (1.0 + tax_rate)
"""

# 1. Inspect the Abstract Syntax Tree
tree = ast.parse(source_code)
print("--- AST DUMP ---")
print(ast.dump(tree, indent=2))

# 2. Compile to a Code Object and disassemble bytecode
code_obj = compile(source_code, filename="<string>", mode="exec")
# Extract the function's code object from the module's constants
func_code = code_obj.co_consts[0]

print("\n--- BYTECODE DISASSEMBLY ---")
dis.dis(func_code)
```

```python
# Simulate a logged-out user
current_user = {"name": "Alex", "is_logged_in": False}


# 1. Define the decorator (just a normal function!)
def login_required(func):
    def wrapper():
        if not current_user["is_logged_in"]:
            print("Access Denied: Please log in first.")
            return  # Stops here; never runs the original function

        # If logged in, run the original function
        return func()

    return wrapper

@login_required
def view_bank_balance():
    print("Your balance is $5,000.")


# --- Let's test it ---
view_bank_balance()
# Output: Access Denied: Please log in first.

current_user["is_logged_in"] = True
view_bank_balance()
# Output: Your balance is $5,000.
```

```python
#!/usr/bin/env python3
"""Diagnostic script to inspect the active Python runtime environment."""

import platform
import sys
from typing import NoReturn


def inspect_runtime() -> dict[str, str | bool]:
    """Gathers key metadata about the executing interpreter."""
    gil_enabled = True
    # Python 3.13+ introduces sys._is_gil_enabled() for free-threaded builds
    if hasattr(sys, "_is_gil_enabled"):
        gil_enabled = sys._is_gil_enabled()

    return {
        "implementation": platform.python_implementation(),
        "version": platform.python_version(),
        "executable": sys.executable,
        "architecture": platform.machine(),
        "gil_enabled": gil_enabled,
    }


def main() -> int:
    """Primary entry point returning an OS-compatible exit status code."""
    try:
        info = inspect_runtime()
        for key, value in info.items():
            print(f"{key:<16}: {value}")
        return 0
    except Exception as exc:
        print(f"Fatal runtime inspection error: {exc}", file=sys.stderr)
        return 1


def run() -> NoReturn:
    """Wrapper to exit cleanly with the integer status code from main()."""
    main() #sys.exit(main())


if __name__ == "__main__":
    run()
```
