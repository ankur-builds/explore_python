```python
# Numeric conversions (invoking the int and float constructors)
x = int("42")             # 42 (parses base-10 string into a new int object)
z = int("1010", base=2)   # 10 (supports bases 2–36, or base=0 to auto-detect 0x/0b/0o prefixes)
c = int(3.99)             # 3  (truncates toward zero—does NOT round! Use round(3.99) for 4)

q = float("inf")          # IEEE 754 positive infinity
s = float("nan")          # IEEE 754 Not-a-Number
# Watch out: per IEEE 754 rules, NaN is never equal to anything—even itself!
print(float("nan") == float("nan"))  # False (use math.isnan(x) instead)
print (x, z, c, q, s)
```

```python
raw_bytes = "héllo".encode("utf-8")     # b'h\xc3\xa9llo' (note 'é' takes 2 bytes in UTF-8)
decoded_str = raw_bytes.decode("utf-8") # 'héllo'

print (raw_bytes)
print (decoded_str)
```

```python
from enum import IntEnum

class Status(IntEnum):
    OK = 200
    NOT_FOUND = 404


def check_status_buggy(code: int) -> str:
    match code:
        case Status.OK:  # ⚠️ BUG! `OK` has NO dot—Python treats it as a new capture variable!
            return f"Matched OK (and silently overwrote OK = {OK})"
        case Status.NOT_FOUND:  # SyntaxError: unreachable! Previous pattern already matched everything
            return "Not Found"
        case _ :
            return "5xx error"

check_status_buggy(500)
```

```python
tasks = ["build", "test", "deploy"]

# ❌ Clunky C/Java style:
for i in range(len(tasks)):
    print(i, tasks[i])

# ✅ Clean Pythonic style (can even start counting at 1!):
for idx, task in enumerate(tasks, start=1):
    print(f"Step {idx}: {task}")
```

```python
import sys

lst = []
prev_size = sys.getsizeof(lst)
print(f"Start: len={len(lst)}, bytes={prev_size}")

for i in range(20):
  lst.append(i)
  curr_size = sys.getsizeof(lst)
  if curr_size != prev_size:
    added_slots = (curr_size - prev_size) // 8
    print(
        f"len={len(lst):2d} -> Resized! Allocated {added_slots} extra pointer"
        f" slots (total {curr_size} bytes)"
    )
    prev_size = curr_size
```
